### Document Structure


```python
from langchain_core.documents import Document
```

```python
doc = Document(
    page_content="""Retrieval-Augmented Generation (RAG) combines document retrieval with language generation. Before answering a question, the system searches a knowledge base for relevant passages and provides those passages as context to the language model. This helps produce answers that are grounded in the source documents and easier to verify.""",
    metadata={
        "source": "rag_introduction.txt",
        "title": "Introduction to Retrieval-Augmented Generation",
        "category": "artificial intelligence",
        "language": "en",
        "page": 1,
        "date_created":"01-03-2026"
    }
)
```

```python
# Create a simple text file
import os
from pathlib import Path

text_directory = Path("../data/text_files")
text_directory.mkdir(parents=True, exist_ok=True)

sample_file = text_directory / "rag_introduction.txt"
sample_file.write_text(
    """Retrieval-Augmented Generation (RAG) combines document retrieval with language generation.

Before answering a question, a RAG system searches a knowledge base for relevant passages and provides them as context to a language model. This helps produce answers that are grounded in source documents and easier to verify.
""",
    encoding="utf-8",
)

print(f"Created: {sample_file}")
```

```python
from langchain_community.document_loaders import TextLoader
loader=TextLoader("../data/text_files/rag_introduction.txt", encoding="utf-8")
document=loader.load()
document
```
